```python
import pandas as pd

# Load raw dataset
df = pd.read_csv("C:\\Users\\Santos\\Desktop\\pipeline project\\best_selling_albums_messy.csv") 

# Preview data
df.head()
```

```python
df.isnull().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Standardize column names
df.columns = df.columns.str.lower().str.replace(" ", "_")

# Fix text columns
df['artist'] = df['artist'].str.strip().str.title()
df['album'] = df['album'].str.strip().str.title()

# Convert sales column to numeric (example)
df['sales_millions'] = (
    df['sales_millions']
    .astype(str)
    .str.replace(",", "")
    .astype(float)
)

# Handle missing values
df = df.dropna()
```

```python
file_path = 'my_data.csv'
```

```python
df.to_csv('my_data.csv', index=False)
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+mysqlconnector://root:Benedict2005@localhost:3306/music_pipeline"
)

df.to_sql(
    name="best_selling_albums",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
top_albums = pd.read_sql("""
    SELECT album, artist, sales_millions
    FROM best_selling_albums
    ORDER BY sales_millions DESC
""", engine)

top_albums.to_excel("final_analysis.xlsx", index=False)
```
